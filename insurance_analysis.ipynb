```python
insurance.groupby('smoker')['charges'].mean()
```

```python
smoker_charges = insurance.groupby('smoker')['charges'].mean()

print(smoker_charges)
```

```python
import matplotlib.pyplot as plt

smoker_charges.plot(
    kind='bar',
    color=['green', 'red']
)

plt.title('Average Insurance Charges')
plt.ylabel('Charges ($)')
plt.show()
```

```python
insurance.groupby('region')['charges'].mean()
```

```python
regional_charges = insurance.groupby('region')['charges'].mean()

regional_charges.sort_values(ascending=False)
```

```python
regional_charges.plot(kind='bar')

plt.title('Average Charges by Region')
plt.ylabel('Charges ($)')
plt.xlabel('Region')

plt.show()
```
